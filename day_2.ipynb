Area of Rectangle:

```python
l=10
b=7
area_rectangle=l*b
print('Area of Rectangle is',area_rectangle)
print(f'Area of rectangle is {area_rectangle:.0f}')
```

Area of Circle:

```python
r=10.5
area_circle=3.14*r*r
print(f'Area of Circle with radius {r} is {area_circle}')
```

##Volume of cylinder :3.14*(r**2)*h

```python
r=10.5
h=50
vol_cylinder=3.14*(r**2)*h
print(f'Volume of Cylinder is {vol_cylinder} with radius {r} and height is {h}')
```
